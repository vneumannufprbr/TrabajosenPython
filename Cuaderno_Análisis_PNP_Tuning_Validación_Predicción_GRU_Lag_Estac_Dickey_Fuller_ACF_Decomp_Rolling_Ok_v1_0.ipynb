```python
# -*- coding: utf-8 -*-
"""
Cuaderno de Análisis y Predicción (GRU con ACF, Decomp y Rolling,
Autoajuste y Gráficos)

Versión 3.0 - Lags Estacionales Configurables

Objetivo: Permitir la configuración de lags estacionales específicos para cada
          tipo de incidente, mejorando la adaptabilidad del modelo a los patrones
          detectados en el análisis exploratorio (ACF, PACF, Decomp).
"""

!pip install tensorflow keras-tuner

import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import GRU, Dense, Dropout
from sklearn.preprocessing import MinMaxScaler
import keras_tuner as kt
import holidays
import os
import joblib
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from sklearn.metrics import mean_squared_error, r2_score
from statsmodels.tsa.stattools import adfuller
import matplotlib.ticker as mticker
import warnings

# --- CONFIGURACIÓN DE PARÁMETROS ---
LAGS_TO_PLOT = 40
look_back = 30

# Diccionario para configurar el tamaño de la ventana de la media móvil
ROLLING_WINDOW_SIZES = {
    'Abigeato': 7,
    'Coaccion sexual': 7,
    'Homicidio culposo en accidente de transito': 7,
    'Homicidio doloso': 30, # Ejemplo de cambio
    'Hurto': 7,
    'Hurto agravado': 7,
    'Lesion': 7,
    'Lesion culposa en accidente de transito': 7,
    'Robo agravado de motocicletas': 14, # Ejemplo de cambio
    'Robo agravado de objetos': 7,
    'Robo agravado de vehiculos': 7,
    'Robo de motocicletas': 14, # Ejemplo de cambio
    'Robo de objetos': 7,
    'Robo de vehiculos': 7,
}

# --- ¡NUEVO! ---
# Diccionario para configurar el lag estacional específico por incidente.
# Puedes ajustar estos valores según los resultados de tu análisis ACF/PACF.
SEASONAL_LAGS = {
    'Abigeato': 28,
    'Coaccion sexual': 28,
    'Homicidio culposo en accidente de transito': 7, # Estacionalidad semanal detectada
    'Homicidio doloso': 365,
    'Hurto': 7, # Estacionalidad semanal detectada
    'Hurto agravado': 365,
    'Lesion': 365,
    'Lesion culposa en accidente de transito': 365,
    'Robo agravado de motocicletas': 365,
    'Robo agravado de objetos': 365,
    'Robo agravado de vehiculos': 365,
    'Robo de motocicletas': 365,
    'Robo de objetos': 365,
    'Robo de vehiculos': 365,
}


# Inicializar diccionarios para almacenar modelos e hiperparámetros
if 'best_models' not in locals():
    best_models = {}
if 'best_hps_dict' not in locals():
    best_hps_dict = {}

# Definir el directorio para los modelos validados
validated_models_dir = "modelos_validados"
os.makedirs(validated_models_dir, exist_ok=True)


# --- Carga de Datos ---
if 'df_granular' not in locals() or df_granular.empty:
    print("df_granular no encontrado. Intentando recargar los datos.")
    try:
        df_incidents = pd.read_csv('incidentes_ampliados_7_anios_completo.csv', encoding='utf-8')
        df_incidents['fecha'] = pd.to_datetime(df_incidents['fecha'])
        df_pivot = pd.pivot_table(df_incidents, index='fecha', columns='tipo_incidente', aggfunc='size', fill_value=0)
        date_rng = pd.date_range(start=df_pivot.index.min(), end=df_pivot.index.max(), freq='D')
        df_granular = df_pivot.reindex(date_rng, fill_value=0)
        print(f"Datos cargados exitosamente con {len(df_granular.columns)} tipos de incidentes.")
    except FileNotFoundError:
        print("\nERROR: El archivo 'incidentes_ampliados_7_anios_completo.csv' no se encontró. No se puede continuar.")
        df_granular = pd.DataFrame()


# --- Definición de Funciones ---

# <--- MODIFICADO ---
def crear_caracteristicas(dataframe, target_column, window_size, seasonal_lag):
    """Crea características de series temporales, incluyendo un lag estacional configurable."""
    df_copy = dataframe.copy()
    df_copy['dia_de_la_semana'] = df_copy.index.dayofweek
    df_copy['mes'] = df_copy.index.month
    df_copy['anio'] = df_copy.index.year
    df_copy['dia_del_anio'] = df_copy.index.dayofyear
    df_copy['semana_del_anio'] = df_copy.index.isocalendar().week.astype(int)

    try:
        if not df_copy.index.empty:
            py_holidays = holidays.Paraguay(years=df_copy.index.year.unique())
            df_copy['es_feriado'] = df_copy.index.isin(py_holidays).astype(int)
        else:
            df_copy['es_feriado'] = 0
    except Exception as e:
        print(f"Advertencia: No se pudieron generar las características de feriados. Error: {e}")
        df_copy['es_feriado'] = 0

    if target_column in df_copy.columns:
        df_copy['lag1'] = df_copy[target_column].shift(1)
        df_copy['lag7'] = df_copy[target_column].shift(7)
        # --- ¡NUEVO! --- Usa el lag estacional configurable
        df_copy[f'lag_{seasonal_lag}'] = df_copy[target_column].shift(seasonal_lag)
        df_copy['rolling_mean'] = df_copy[target_column].rolling(window=window_size).mean()
    else:
        df_copy['lag1'] = np.nan
        df_copy['lag7'] = np.nan
        df_copy[f'lag_{seasonal_lag}'] = np.nan
        df_copy['rolling_mean'] = np.nan

    return df_copy

# <--- MODIFICADO ---
def preprocesar_para_gru(dataframe, target_column, window_size, seasonal_lag, look_back):
    """Procesa datos, normaliza y crea secuencias para el modelo GRU."""
    df_with_features = crear_caracteristicas(dataframe, target_column, window_size, seasonal_lag)
    df_processed = df_with_features.dropna()

    if df_processed.empty:
        print(f"Advertencia: El dataframe para '{target_column}' está vacío después de eliminar NaNs.")
        return None

    features = df_processed.drop(columns=[target_column])
    target = df_processed[[target_column]]

    feature_scaler = MinMaxScaler(feature_range=(0, 1))
    scaled_features = feature_scaler.fit_transform(features)

    target_scaler = MinMaxScaler(feature_range=(0, 1))
    scaled_target = target_scaler.fit_transform(target)

    X_sequence, y_sequence, sequence_dates = [], [], []
    for i in range(len(scaled_features) - look_back):
        X_sequence.append(scaled_features[i:(i + look_back)])
        y_sequence.append(scaled_target[i + look_back])
        sequence_dates.append(df_processed.index[i + look_back])

    if not X_sequence:
        print(f"Advertencia: No se pudieron crear secuencias para '{target_column}'. Datos insuficientes.")
        return None

    print(f"  - Formas de secuencias GRU (X, y): {np.array(X_sequence).shape}, {np.array(y_sequence).shape}")

    return {
        'X_sequence': np.array(X_sequence),
        'y_sequence': np.array(y_sequence),
        'feature_scaler': feature_scaler,
        'target_scaler': target_scaler,
        'sequence_dates': pd.DatetimeIndex(sequence_dates)
    }

def build_gru_model(hp, look_back, n_features):
    """Construye un modelo GRU con hiperparámetros ajustables."""
    model = Sequential()
    model.add(GRU(units=hp.Int('units', min_value=32, max_value=128, step=32),
                  return_sequences=False,
                  input_shape=(look_back, n_features)))
    if hp.Boolean('use_dropout'):
        model.add(Dropout(hp.Float('dropout', min_value=0.1, max_value=0.5, step=0.1)))
    model.add(Dense(1))
    model.compile(optimizer='adam', loss='mean_squared_error')
    return model

def analizar_series_temporales(dataframe, lags):
    """Genera gráficos ACF, PACF, descomposición estacional y realiza el test de Dickey-Fuller."""
    print("\n" + "="*70 + "\nINICIANDO MODO ANÁLISIS\n" + "="*70)
    if dataframe.empty:
        print("El DataFrame está vacío. No se puede realizar el análisis.")
        return
    for incidente in dataframe.columns:
        print(f"\n--- Análisis para: '{incidente}' ---")
        series = dataframe[incidente]
        if series.sum() > 20:
            fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(15, 8))
            plot_acf(series, ax=ax1, lags=lags)
            ax1.set_title(f'Función de Autocorrelación (ACF) para {incidente}')
            plot_pacf(series, ax=ax2, lags=lags)
            ax2.set_title(f'Función de Autocorrelación Parcial (PACF) para {incidente}')
            plt.tight_layout()
            plt.show()
        else:
            print("Datos insuficientes para un análisis de autocorrelación significativo.")
        try:
            result = seasonal_decompose(series, model='additive', period=365)
            fig = result.plot()
            fig.set_size_inches(15, 8)
            plt.suptitle(f'Descomposición Estacional para {incidente}', y=1.02)
            plt.tight_layout()
            plt.show()
        except Exception as e:
            print(f"No se pudo realizar la descomposición estacional para '{incidente}': {e}")
        try:
            result_adf = adfuller(series.dropna())
            print(f"--- Test de Dickey-Fuller Aumentado para: '{incidente}' ---")
            print(f'  Estadística ADF: {result_adf[0]:.4f}')
            print(f'  Valor p: {result_adf[1]:.4f}')
            if result_adf[1] <= 0.05:
                print(f'  Conclusión: La serie parece ser ESTACIONARIA (se rechaza la hipótesis nula).')
            else:
                print(f'  Conclusión: La serie parece NO ser ESTACIONARIA (no se puede rechazar la hipótesis nula).')
        except Exception as e:
            print(f"No se pudo realizar el Test de Dickey-Fuller para '{incidente}': {e}")

# <--- MODIFICADO ---
def entrenar_y_guardar_modelo(dataframe, target_column, rolling_window_sizes, seasonal_lags, look_back, best_hps_dict, save_dir):
    """Entrena un modelo GRU con TODOS los datos disponibles y lo guarda."""
    window_size = rolling_window_sizes.get(target_column, 7)
    seasonal_lag = seasonal_lags.get(target_column, 365) # Obtener lag del nuevo diccionario
    print(f"\n--- Entrenando modelo de producción para '{target_column}' (Lag Estacional: {seasonal_lag}) ---")

    processed_data = preprocesar_para_gru(dataframe[[target_column]], target_column, window_size, seasonal_lag, look_back)
    if processed_data is None:
        print(f"No se entrenará ni guardará el modelo para '{target_column}' por falta de datos.")
        return

    X, y, feature_scaler, target_scaler = processed_data['X_sequence'], processed_data['y_sequence'], processed_data['feature_scaler'], processed_data['target_scaler']
    n_features = X.shape[2]

    best_hps = best_hps_dict.get(target_column)
    if not best_hps:
        print(f"No se encontraron hiperparámetros para '{target_column}'. No se puede entrenar.")
        return

    model = build_gru_model(kt.HyperParameters.from_config(best_hps), look_back, n_features)
    print(f"  - Entrenando el mejor modelo para '{target_column}' en todos los datos disponibles...")
    model.fit(X, y, epochs=100, batch_size=32, verbose=0)

    best_models[target_column] = model
    print(f"  - Modelo entrenado y almacenado para '{target_column}'.")

    model_path = os.path.join(save_dir, f"best_gru_model_{target_column.lower().replace(' ', '_')}.keras")
    model.save(model_path)
    joblib.dump(feature_scaler, os.path.join(save_dir, f"feature_scaler_{target_column.lower().replace(' ', '_')}.joblib"))
    joblib.dump(target_scaler, os.path.join(save_dir, f"target_scaler_{target_column.lower().replace(' ', '_')}.joblib"))
    print(f"  - Modelo y scalers para '{target_column}' guardados en '{save_dir}'")

# <--- MODIFICADO ---
def validar_modelo(dataframe, target_column, rolling_window_sizes, seasonal_lags, look_back, best_models):
    """Valida un modelo GRU pre-entrenado usando un conjunto de prueba."""
    window_size = rolling_window_sizes.get(target_column, 7)
    seasonal_lag = seasonal_lags.get(target_column, 365)
    print("-" * 70 + f"\nVALIDANDO MODELO GRU PARA: '{target_column}'")

    processed_data = preprocesar_para_gru(dataframe[[target_column]], target_column, window_size, seasonal_lag, look_back)
    if processed_data is None:
        print(f"Datos insuficientes para validar '{target_column}'.")
        return None, None, None

    X, y, dates, target_scaler = processed_data['X_sequence'], processed_data['y_sequence'], processed_data['sequence_dates'], processed_data['target_scaler']

    fecha_corte = dates.max() - pd.DateOffset(months=3)
    cut_off_idx = dates.searchsorted(fecha_corte)
    if cut_off_idx >= len(dates) - 1:
        print(f"No hay suficientes datos de prueba para '{target_column}'.")
        return None, None, None

    X_test, y_test, test_dates = X[cut_off_idx:], y[cut_off_idx:], dates[cut_off_idx:]
    model = best_models.get(target_column)
    if model is None:
        print(f"No se encontró un modelo entrenado para '{target_column}'.")
        return None, None, None

    y_pred_scaled = model.predict(X_test, verbose=0)
    y_test_actual = target_scaler.inverse_transform(y_test)
    y_pred_actual = target_scaler.inverse_transform(y_pred_scaled)

    rmse = np.sqrt(mean_squared_error(y_test_actual, y_pred_actual))
    r2 = r2_score(y_test_actual, y_pred_actual)
    print(f"\nRESULTADOS DE VALIDACIÓN:\n  - RMSE: {rmse:.2f}\n  - R2 Score: {r2:.2f}")

    plt.figure(figsize=(15, 6))
    plt.plot(dataframe.index, dataframe[target_column], label='Histórico Completo', color='gray', alpha=0.5)
    plt.plot(test_dates, y_test_actual, label='Valores Reales (Prueba)', color='blue')
    plt.plot(test_dates, y_pred_actual, label='Predicciones (Prueba)', color='red', linestyle='--')
    plt.title(f"Validación de Pronóstico GRU para '{target_column}'")
    plt.xlabel('Fecha'), plt.ylabel('Cantidad de Incidentes'), plt.legend(), plt.grid(True)
    plt.show()

    return rmse, r2, None

# <--- MODIFICADO ---
def cargar_y_predecir(full_historical_df, target_column, model_dir, future_dates, rolling_window_sizes, seasonal_lags, look_back):
    """Carga un modelo y predice valores futuros de forma iterativa."""
    try:
        model_path = os.path.join(model_dir, f"best_gru_model_{target_column.lower().replace(' ', '_')}.keras")
        model = tf.keras.models.load_model(model_path)
        feature_scaler = joblib.load(os.path.join(model_dir, f"feature_scaler_{target_column.lower().replace(' ', '_')}.joblib"))
        target_scaler = joblib.load(os.path.join(model_dir, f"target_scaler_{target_column.lower().replace(' ', '_')}.joblib"))
    except Exception as e:
        print(f"Error al cargar el modelo o los scalers para '{target_column}': {e}")
        return None

    window_size = rolling_window_sizes.get(target_column, 7)
    seasonal_lag = seasonal_lags.get(target_column, 365)
    print(f"--- Generando predicciones GRU para '{target_column}' (Lag Estacional: {seasonal_lag}) ---")

    datos_conocidos = full_historical_df.copy()
    future_predictions_actual = []

    for fecha_prediccion in future_dates:
        # Tomar los datos necesarios para la última secuencia, considerando el lag más largo
        max_lag = max(7, seasonal_lag, window_size)
        datos_para_secuencia = datos_conocidos.tail(look_back + max_lag)
        df_features = crear_caracteristicas(datos_para_secuencia, target_column, window_size, seasonal_lag)
        df_features_input = df_features.drop(columns=[target_column]).tail(look_back).fillna(method='ffill').fillna(method='bfill').fillna(0)

        try:
            df_features_input = df_features_input[feature_scaler.feature_names_in_]
            scaled_input = feature_scaler.transform(df_features_input)
        except Exception as e:
            print(f"Error al escalar características para '{target_column}': {e}")
            return None

        input_sequence = np.reshape(scaled_input, (1, look_back, scaled_input.shape[1]))
        pred_scaled = model.predict(input_sequence, verbose=0)
        pred_actual = max(0, round(target_scaler.inverse_transform(pred_scaled)[0][0]))
        future_predictions_actual.append(pred_actual)

        nueva_fila = pd.DataFrame({col: [np.nan] for col in datos_conocidos.columns}, index=[fecha_prediccion])
        nueva_fila[target_column] = pred_actual
        datos_conocidos = pd.concat([datos_conocidos, nueva_fila])

    return np.array(future_predictions_actual)


# --- Lógica Principal de Ejecución ---
if not df_granular.empty:
    user_input = input("Seleccione un modo (analizar, tuning, validar, predecir): ").lower()

    if user_input == 'analizar':
        analizar_series_temporales(df_granular, LAGS_TO_PLOT)

    elif user_input == 'tuning':
        print("\n" + "="*70 + "\nINICIANDO MODO TUNING GRU (KERAS TUNER)\n" + "="*70)
        tuner_dir = "keras_tuner_results"
        for incidente in df_granular.columns:
            print(f"\n--- Ajustando hiperparámetros para: '{incidente}' ---")
            window_size = ROLLING_WINDOW_SIZES.get(incidente, 7)
            seasonal_lag = SEASONAL_LAGS.get(incidente, 365) # <--- MODIFICADO

            processed_data = preprocesar_para_gru(df_granular[[incidente]], incidente, window_size, seasonal_lag, look_back)
            if processed_data is None:
                print(f"Saltando tuning para '{incidente}' por datos insuficientes.")
                continue

            X, y, n_features = processed_data['X_sequence'], processed_data['y_sequence'], processed_data['X_sequence'].shape[2]
            train_size = int(len(X) * 0.8)
            if train_size < 1:
                print(f"Datos insuficientes para dividir en entrenamiento/validación para '{incidente}'.")
                continue

            X_train, y_train, X_val, y_val = X[:train_size], y[:train_size], X[train_size:], y[train_size:]
            tuner = kt.RandomSearch(
                hypermodel=lambda hp: build_gru_model(hp, look_back, n_features),
                objective='val_loss', max_trials=10, executions_per_trial=1,
                directory=tuner_dir, project_name=f'gru_tuning_{incidente.lower().replace(" ", "_")}', overwrite=True
            )
            print(f"  - Iniciando búsqueda de hiperparámetros...")
            tuner.search(X_train, y_train, epochs=50, validation_data=(X_val, y_val), verbose=0)

            best_hps_list = tuner.get_best_hyperparameters(num_trials=1)
            if best_hps_list:
                best_hps_dict[incidente] = best_hps_list[0].get_config()
                print(f"  - Mejores hiperparámetros encontrados para '{incidente}': {best_hps_list[0].values}")
                # <--- MODIFICADO
                entrenar_y_guardar_modelo(df_granular, incidente, ROLLING_WINDOW_SIZES, SEASONAL_LAGS, look_back, best_hps_dict, validated_models_dir)
            else:
                print(f"  - No se encontraron hiperparámetros óptimos para '{incidente}'.")

    elif user_input == 'validar':
        print("\n" + "="*70 + "\nINICIANDO MODO VALIDACIÓN GRU\n" + "="*70)
        if not best_models:
            print("No hay modelos en memoria. Intentando cargar desde el disco...")
            for incidente in df_granular.columns:
                try:
                    model_path = os.path.join(validated_models_dir, f"best_gru_model_{incidente.lower().replace(' ', '_')}.keras")
                    best_models[incidente] = tf.keras.models.load_model(model_path)
                    print(f"  - Modelo para '{incidente}' cargado exitosamente.")
                except:
                    print(f"  - No se encontró un modelo guardado para '{incidente}'.")
        if not best_models:
            print("No hay modelos para validar. Ejecute el modo 'tuning' primero.")
        else:
            for incidente in best_models.keys():
                # <--- MODIFICADO
                validar_modelo(df_granular, incidente, ROLLING_WINDOW_SIZES, SEASONAL_LAGS, look_back, best_models)

    elif user_input == 'predecir':
        print("\n" + "="*70 + "\nINICIANDO MODO PREDICCIÓN GRU\n" + "="*70)
        try:
            num_days = int(input("Ingrese el número de días para predecir: "))
            if num_days <= 0: raise ValueError
        except ValueError:
            print("Entrada inválida. Por favor, ingrese un número entero positivo.")
        else:
            last_date = df_granular.index.max()
            future_dates = pd.date_range(start=last_date + pd.DateOffset(days=1), periods=num_days, freq='D')
            all_predictions = {}
            for incidente in df_granular.columns:
                # <--- MODIFICADO
                predictions = cargar_y_predecir(df_granular, incidente, validated_models_dir, future_dates, ROLLING_WINDOW_SIZES, SEASONAL_LAGS, look_back)
                if predictions is not None:
                    all_predictions[incidente] = predictions
            if all_predictions:
                predictions_df = pd.DataFrame(all_predictions, index=future_dates)
                predictions_df.index.name = 'Fecha'
                print("\n--- Predicciones Futuras ---\n", predictions_df)
                csv_filename = 'pronostico_futuro_gru_lags_config.csv'
                predictions_df.to_csv(csv_filename)
                print(f"\nPredicciones guardadas en '{csv_filename}'")
                for incidente in all_predictions.keys():
                    plt.figure(figsize=(15, 6))
                    last_year_start = df_granular.index.max() - pd.DateOffset(years=1)
                    df_granular.loc[last_year_start:, incidente].plot(label='Histórico (Último Año)', color='blue')
                    predictions_df[incidente].plot(label='Predicciones Futuras', color='red', linestyle='--')
                    plt.title(f"Pronóstico Futuro para '{incidente}'")
                    plt.ylabel('Cantidad de Incidentes'), plt.xlabel('Fecha'), plt.legend(), plt.grid(True)
                    plt.show()
    else:
        print("Modo no reconocido. Por favor, elija entre 'analizar', 'tuning', 'validar' o 'predecir'.")
```
